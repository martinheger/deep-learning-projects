```python
from io import BytesIO
from pypdf import PdfReader
from markitdown import MarkItDown

def pdf_to_mmd(pdf_path, output_path=None):
    # --- 1️⃣ Extract text from the PDF ---
    reader = PdfReader(pdf_path)
    text = "\n\n".join(page.extract_text() or "" for page in reader.pages)

    # --- 2️⃣ Convert text to MultiMarkdown ---
    md = MarkItDown()
    text_stream = BytesIO(text.encode("utf-8"))  # make it look like a file
    result = md.convert(text_stream, file_name="input.txt")

    # --- 3️⃣ Save result as .mmd ---
    output_path = output_path or pdf_path.replace(".pdf", ".mmd")
    with open(output_path, "w", encoding="utf-8") as f:
        f.write(result.text_content)

    print(f"✅ Converted {pdf_path} → {output_path}")

# Run it on your file
pdf_to_mmd("/content/Sb_2009_40_2025-10-01_IZ.pdf")
pdf_to_mmd("/content/Sb_2012_89_2025-07-01_IZ.pdf")
pdf_to_mmd("/content/Sb_2021_283_2025-08-01_IZ.pdf")
```

```python
import glob
import os

def merge_markdown_files(input_folder, output_file):
    # Get all .md and .mmd files in the folder
    files = sorted(
        glob.glob(os.path.join(input_folder, "*.md")) +
        glob.glob(os.path.join(input_folder, "*.mmd"))
    )

    with open(output_file, "w", encoding="utf-8") as outfile:
        for i, file_path in enumerate(files, 1):
            with open(file_path, "r", encoding="utf-8") as infile:
                content = infile.read().strip()
                outfile.write(f"\n\n# File {i}: {os.path.basename(file_path)}\n\n")
                outfile.write(content)
                outfile.write("\n\n" + "="*80 + "\n")  # separator between files

    print(f"✅ Merged {len(files)} files into {output_file}")

# Example usage:
merge_markdown_files("/content", "/content/merged_dataset.md")
```
